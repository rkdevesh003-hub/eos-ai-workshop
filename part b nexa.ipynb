## CELL 1 – Installing + importing

```python
%pip install --upgrade transformers --quiet

import torch
from transformers import AutoTokenizer , AutoModelForCausalLM

device = 'cuda' if torch . cuda . is_available () else 'cpu'
print('using :', device)
```

## CELL 2 – Loading the small open-source LLM

```python
model_id = 'HuggingFaceTB/SmolLM2-135M-Instruct'
# 135M parameters - downloads ~270 MB. Runs on CPU in ~3 seconds per reply .

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id).to(device)

print('model loaded.')
```

## CELL 3 – Defining my persona

```python
persona = """ You are DUSTIN HENDERSON,one of the character from the stranger things series.
You are a nerdy teen who is always obsessed with science,especially physics concepts and also the one who cares a lot about myths and tries to investigate and prove it.
You always speaks in a thoughtful,smart and humourous manner and always want people to respond to you.You are afraid of losing yourloved ones.You are not only smart,you are courageous,determined and a friendly guy
who always clarifies your friends'quieries,respond to their emotions and always motivate them.Your slang should be like an american especially from Hawkins, Indiana.
You always view the world like a physics professor."""
```

##  CELL 4 – The chat function

```python
def chat ( user_message , max_new =120 , temperature =0.7) :
    messages = [
    {'role': 'user', 'content': persona } ,
    {'role': 'user', 'content': user_message } ,
    ]
    prompt = tokenizer . apply_chat_template (
    messages , tokenize = False , add_generation_prompt = True )

    inputs = tokenizer ( prompt , return_tensors ='pt') . to ( device )
    out = model . generate (
    ** inputs ,
    max_new_tokens = max_new ,
    do_sample = True ,
    temperature = temperature ,
    top_p =0.9 ,
    pad_token_id = tokenizer . eos_token_id )

    reply = tokenizer . decode (
    out [0][ inputs ['input_ids']. shape [1]:] ,
    skip_special_tokens = True )
    return reply . strip ()

# Test it!
print ( chat (user_message='What do you think about my new iPhone ?') )

```
